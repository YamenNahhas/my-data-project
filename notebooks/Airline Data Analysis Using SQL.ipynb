```python
#join flight information with revenue
sql_query = '''SELECT 
    f.flight_id,
    f.departure_airport,
    f.arrival_airport,
    sum(tf.amount) AS revenue
FROM ticket_flights tf
JOIN flights f 
    ON tf.flight_id = f.flight_id
GROUP BY f.flight_id'''
df_join_flight_info_revenue = pd.read_sql_query(sql_query, conn)
display(df_join_flight_info_revenue)
```

```python
#join aircraft code with revenue
sql_query ="""SELECT 
    a.aircraft_code,
    SUM(tf.amount) AS revenue
FROM ticket_flights tf
JOIN flights f 
    ON tf.flight_id = f.flight_id
JOIN aircrafts_data a 
    ON f.aircraft_code = a.aircraft_code
GROUP BY a.aircraft_code"""
df_join_aircraft_code_revenue=pd.read_sql_query(sql_query, conn)
display(df_join_aircraft_code_revenue)
```

```python
sql_query = """SELECT 
    b.book_ref,
    b.total_amount,
    COUNT(tf.ticket_no) AS tickets
FROM bookings b
JOIN tickets t 
    ON b.book_ref = t.book_ref
JOIN ticket_flights tf 
    ON t.ticket_no = tf.ticket_no
GROUP BY b.book_ref"""

df_total_amount_tickets = pd.read_sql_query(sql_query, conn)
display(df_total_amount_tickets)
```

```python
#calculation of load factor
sql_query = '''with occupied AS (
    SELECT 
        flight_id,
        COUNT(*) AS occupied_seats
    FROM boarding_passes
    GROUP BY flight_id
),

total AS (
    SELECT 
        f.flight_id,
        COUNT(s.seat_no) AS total_seats
    FROM flights f
    JOIN seats s 
        ON f.aircraft_code = s.aircraft_code
    GROUP BY f.flight_id
)

SELECT 
    o.flight_id,
    o.occupied_seats,
    t.total_seats,
    o.occupied_seats * 1.0 / t.total_seats AS load_factor
FROM occupied o
JOIN total t 
    ON o.flight_id = t.flight_id'''

df_load_factor = pd.read_sql_query(sql_query, conn)
display(df_load_factor)
```

```python
#flights with load factor more than 85%
df_load_fctor_over_85 = df_load_factor[df_load_factor['load_factor']>0.8]
display(df_load_fctor_over_85)
```

```python
#calculation of revenue per flight
sql_query = '''SELECT 
    flight_id,
    SUM(amount) AS revenue
FROM ticket_flights
GROUP BY flight_id'''

df_revenue_per_flight = pd.read_sql_query(sql_query, conn)
display(df_revenue_per_flight)
```

```python
#comparison of Business vs Economy
sql_query = '''WITH occupied AS (
    SELECT 
        tf.fare_conditions,
        COUNT(*) AS occupied_seats
    FROM boarding_passes bp
    JOIN ticket_flights tf
        ON bp.ticket_no = tf.ticket_no
       AND bp.flight_id = tf.flight_id
    GROUP BY tf.fare_conditions
),

pricing AS (
    SELECT 
        fare_conditions,
        AVG(amount) AS avg_price,
        COUNT(*) AS total_tickets
    FROM ticket_flights
    GROUP BY fare_conditions
)

SELECT 
    p.fare_conditions,
    o.occupied_seats,
    p.total_tickets,
    p.avg_price
FROM pricing p
LEFT JOIN occupied o
    ON p.fare_conditions = o.fare_conditions'''     
df_fare_condition_comparison = pd.read_sql_query(sql_query, conn)
display(df_fare_condition_comparison)
```

```python
#calculation of revenue per seat
sql_query = '''WITH revenue AS (
    SELECT 
        flight_id,
        SUM(amount) AS total_revenue
    FROM ticket_flights
    GROUP BY flight_id
),

seats_total AS (
    SELECT 
        f.flight_id,
        COUNT(s.seat_no) AS total_seats
    FROM flights f
    JOIN seats s 
        ON f.aircraft_code = s.aircraft_code
    GROUP BY f.flight_id
)

SELECT 
    r.flight_id,
    r.total_revenue,
    s.total_seats,
    r.total_revenue * 1.0 / s.total_seats AS revenue_per_seat
FROM revenue r
JOIN seats_total s 
    ON r.flight_id = s.flight_id
ORDER BY revenue_per_seat DESC'''

df_revenue_per_seat = pd.read_sql_query(sql_query, conn)
display(df_revenue_per_seat)
```

```python
#binding revenue per seat with the type of aircraft
sql_query = '''
WITH revenue AS (
    SELECT 
        f.flight_id,
        f.aircraft_code,
        SUM(tf.amount) AS total_revenue
    FROM ticket_flights tf
    JOIN flights f ON tf.flight_id = f.flight_id
    GROUP BY f.flight_id, f.aircraft_code
),

seats_total AS (
    SELECT 
        aircraft_code,
        COUNT(seat_no) AS total_seats
    FROM seats
    GROUP BY aircraft_code
)

SELECT 
    r.aircraft_code,
    AVG(r.total_revenue * 1.0 / s.total_seats) AS avg_revenue_per_seat,
    a.model
FROM revenue r
JOIN seats_total s 
    ON r.aircraft_code = s.aircraft_code
JOIN  aircrafts_data a
    ON r.aircraft_code = a.aircraft_code
GROUP BY r.aircraft_code, a.model
ORDER BY avg_revenue_per_seat DESC'''
df_revenue_per_seat_aircraft = pd.read_sql_query(sql_query, conn)
display(df_revenue_per_seat_aircraft)
```

```python
#anaylying of time of flights
sql_query = '''SELECT 
    strftime('%Y-%m-%d', substr(f.scheduled_departure, 1, 19)) AS year_month_day,
    COUNT(tf.ticket_no) AS total_tickets,
    SUM(tf.amount) AS total_revenue
FROM ticket_flights tf
JOIN flights f 
    ON tf.flight_id = f.flight_id
GROUP BY year_month_day
ORDER BY year_month_day'''
df_departure_time = pd.read_sql_query(sql_query, conn)
display(df_departure_time)
```

```python
#analysing of time in weekdays
sql_query = '''SELECT 
    strftime('%w', substr(f.scheduled_departure, 1, 19)) AS weekday_number,
    
    CASE strftime('%w', substr(f.scheduled_departure, 1, 19))
        WHEN '0' THEN 'Sonntag'
        WHEN '1' THEN 'Montag'
        WHEN '2' THEN 'Dienstag'
        WHEN '3' THEN 'Mittwoch'
        WHEN '4' THEN 'Donnerstag'
        WHEN '5' THEN 'Freitag'
        WHEN '6' THEN 'Samstag'
    END AS weekday_name,

    COUNT(tf.ticket_no) AS total_tickets,
    SUM(tf.amount) AS total_revenue,
    COUNT(DISTINCT f.flight_id) AS total_flights

FROM flights f
JOIN ticket_flights tf
    ON f.flight_id = tf.flight_id

GROUP BY weekday_number
ORDER BY weekday_number'''
df_departure_week_days = pd.read_sql_query(sql_query, conn)
display(df_departure_week_days)
```

```python
#anaylying of time of flights in hours
sql_query = '''SELECT 
    strftime('%H', substr(f.scheduled_departure, 1, 19)) AS hour,
    COUNT(tf.ticket_no) AS total_tickets,
    SUM(tf.amount) AS total_revenue
FROM ticket_flights tf
JOIN flights f 
    ON tf.flight_id = f.flight_id
GROUP BY hour'''
df_departure_time_hour= pd.read_sql_query(sql_query, conn)
display(df_departure_time_hour)
```
