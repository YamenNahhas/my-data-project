```python
#calculation of load factor
sql_query = '''with occupied AS (
    SELECT 
        flight_id,
        COUNT(*) AS occupied_seats
    FROM boarding_passes
    GROUP BY flight_id
),

total AS (
    SELECT 
        f.flight_id,
        COUNT(s.seat_no) AS total_seats
    FROM flights f
    JOIN seats s 
        ON f.aircraft_code = s.aircraft_code
    GROUP BY f.flight_id
)

SELECT 
    o.flight_id,
    o.occupied_seats,
    t.total_seats,
    o.occupied_seats * 1.0 / t.total_seats AS load_factor
FROM occupied o
JOIN total t 
    ON o.flight_id = t.flight_id'''

df_load_factor = pd.read_sql_query(sql_query, conn)
display(df_load_factor)
```

```python
#flights with load factor more than 85%
df_load_fctor_over_85 = df_load_factor[df_load_factor['load_factor']>0.8]
print('flights with load factor more than 85%')
display(df_load_fctor_over_85)
```

```python
#analysing of load factor with departure airport
sql_query = '''WITH occupied AS (
    SELECT 
        f.flight_id,
        f.departure_airport,
        COUNT(bp.seat_no) AS occupied_seats
    FROM flights f
    LEFT JOIN boarding_passes bp 
        ON f.flight_id = bp.flight_id
    GROUP BY f.flight_id, f.departure_airport
),

capacity AS (
    SELECT 
        f.flight_id,
        COUNT(s.seat_no) AS total_seats
    FROM flights f
    JOIN seats s 
        ON f.aircraft_code = s.aircraft_code
    GROUP BY f.flight_id
)

SELECT 
    o.departure_airport,
    COUNT(o.flight_id) AS total_flights,
    SUM(o.occupied_seats) AS total_occupied_seats,
    SUM(c.total_seats) AS total_available_seats,
    
    SUM(o.occupied_seats) * 1.0 / SUM(c.total_seats) AS load_factor

FROM occupied o
JOIN capacity c 
    ON o.flight_id = c.flight_id

GROUP BY o.departure_airport
ORDER BY load_factor DESC'''
df_load_factor_departure_airport = pd.read_sql_query(sql_query, conn)
df_load_factor_departure_airport.to_sql(
    name="load_factor_departure_airport",
    con=conn,
    if_exists="replace",
    index=False
)
display(df_load_factor_departure_airport)
```

```python
#analysing of load factor with arrival airport
sql_query = '''WITH occupied AS (
    SELECT 
        f.flight_id,
        f.arrival_airport,
        COUNT(bp.seat_no) AS occupied_seats
    FROM flights f
    LEFT JOIN boarding_passes bp 
        ON f.flight_id = bp.flight_id
    GROUP BY f.flight_id, f.departure_airport
),

capacity AS (
    SELECT 
        f.flight_id,
        COUNT(s.seat_no) AS total_seats
    FROM flights f
    JOIN seats s 
        ON f.aircraft_code = s.aircraft_code
    GROUP BY f.flight_id
)

SELECT 
    o.arrival_airport,
    COUNT(o.flight_id) AS total_flights,
    SUM(o.occupied_seats) AS total_occupied_seats,
    SUM(c.total_seats) AS total_available_seats,
    
    SUM(o.occupied_seats) * 1.0 / SUM(c.total_seats) AS load_factor

FROM occupied o
JOIN capacity c 
    ON o.flight_id = c.flight_id

GROUP BY o.arrival_airport
ORDER BY load_factor DESC'''
df_load_factor_arrival_airport = pd.read_sql_query(sql_query, conn)
df_load_factor_arrival_airport.to_sql(
    name="load_factor_arrival_airport",
    con=conn,
    if_exists="replace",
    index=False
)
display(df_load_factor_arrival_airport)
```

```python
#comparison of Business vs Economy by calculating of average price
sql_query = '''WITH occupied AS (
    SELECT 
        tf.fare_conditions,
        COUNT(*) AS occupied_seats
    FROM boarding_passes bp
    JOIN ticket_flights tf
        ON bp.ticket_no = tf.ticket_no
       AND bp.flight_id = tf.flight_id
    GROUP BY tf.fare_conditions
),

pricing AS (
    SELECT 
        fare_conditions,
        AVG(amount) AS avg_price,
        COUNT(*) AS total_tickets
    FROM ticket_flights
    GROUP BY fare_conditions
)

SELECT 
    p.fare_conditions,
    o.occupied_seats,
    p.total_tickets,
    p.avg_price
FROM pricing p
LEFT JOIN occupied o
    ON p.fare_conditions = o.fare_conditions'''     
df_fare_condition_comparison = pd.read_sql_query(sql_query, conn)
display(df_fare_condition_comparison)
```

```python
sql_query = '''WITH occupied AS (
    SELECT 
        tf.fare_conditions,
        COUNT(*) AS occupied_seats
    FROM boarding_passes bp
    JOIN ticket_flights tf
        ON bp.ticket_no = tf.ticket_no
       AND bp.flight_id = tf.flight_id
    GROUP BY tf.fare_conditions
)

SELECT 
    tf.fare_conditions,
    tf.amount,
    o.occupied_seats
FROM ticket_flights tf
LEFT JOIN occupied o
    ON tf.fare_conditions = o.fare_conditions'''

df_fare_condition = pd.read_sql_query(sql_query, conn)
display(df_fare_condition)
median_table = df_fare_condition.groupby('fare_conditions').agg(
        occupied_seats=('occupied_seats', 'first'),
        total_tickets=('amount', 'count'),
        median_price=('amount', 'median')).round(3).reset_index()
print('comparison of fare conditions by calculationg the median of prices for each group')
display(median_table)
```

```python
#calculation of revenue per seat
sql_query = '''WITH revenue AS (
    SELECT 
        flight_id,
        SUM(amount) AS total_revenue
    FROM ticket_flights
    GROUP BY flight_id
),

seats_total AS (
    SELECT 
        f.flight_id,
        COUNT(s.seat_no) AS total_seats
    FROM flights f
    JOIN seats s 
        ON f.aircraft_code = s.aircraft_code
    GROUP BY f.flight_id
)

SELECT 
    r.flight_id,
    r.total_revenue,
    s.total_seats,
    r.total_revenue * 1.0 / s.total_seats AS revenue_per_seat
FROM revenue r
JOIN seats_total s 
    ON r.flight_id = s.flight_id
ORDER BY revenue_per_seat DESC'''

df_revenue_per_seat = pd.read_sql_query(sql_query, conn)
display(df_revenue_per_seat)
```

```python
#binding revenue per seat with the type of aircraft
sql_query = '''
WITH revenue AS (
    SELECT 
        f.flight_id,
        f.aircraft_code,
        SUM(tf.amount) AS total_revenue
    FROM ticket_flights tf
    JOIN flights f ON tf.flight_id = f.flight_id
    GROUP BY f.flight_id, f.aircraft_code
),

seats_total AS (
    SELECT 
        aircraft_code,
        COUNT(seat_no) AS total_seats
    FROM seats
    GROUP BY aircraft_code
)

SELECT 
    r.aircraft_code,
    AVG(r.total_revenue * 1.0 / s.total_seats) AS avg_revenue_per_seat,
    a.model
FROM revenue r
JOIN seats_total s 
    ON r.aircraft_code = s.aircraft_code
JOIN  aircrafts_data a
    ON r.aircraft_code = a.aircraft_code
GROUP BY r.aircraft_code, a.model
ORDER BY avg_revenue_per_seat DESC'''
df_revenue_per_seat_aircraft = pd.read_sql_query(sql_query, conn)
print('binding revenue per seat with the type of aircraft')
display(df_revenue_per_seat_aircraft)
```

```python
#analyzing of time of departure in weekdays
sql_query = '''SELECT 
    strftime('%w', substr(f.scheduled_departure, 1, 19)) AS weekday_number,
    
    CASE strftime('%w', substr(f.scheduled_departure, 1, 19))
        WHEN '0' THEN 'Sonntag'
        WHEN '1' THEN 'Montag'
        WHEN '2' THEN 'Dienstag'
        WHEN '3' THEN 'Mittwoch'
        WHEN '4' THEN 'Donnerstag'
        WHEN '5' THEN 'Freitag'
        WHEN '6' THEN 'Samstag'
    END AS weekday_name,

    COUNT(tf.ticket_no) AS total_tickets,
    SUM(tf.amount) AS total_revenue,
    COUNT(DISTINCT f.flight_id) AS total_flights

FROM flights f
JOIN ticket_flights tf
    ON f.flight_id = tf.flight_id

GROUP BY weekday_number
ORDER BY weekday_number'''
df_departure_week_days = pd.read_sql_query(sql_query, conn)
print('analyzing of time of departure in weekdays')
display(df_departure_week_days)
```

```python
#calculation of median booking lead time
sql_query = '''SELECT 
    (
        julianday(substr(f.scheduled_departure, 1, 19)) - 
        julianday(substr(b.book_date, 1, 19))
    ) AS booking_lead_time_days
FROM ticket_flights tf
JOIN tickets t 
    ON tf.ticket_no = t.ticket_no
JOIN bookings b
    ON t.book_ref = b.book_ref
JOIN flights f 
    ON tf.flight_id = f.flight_id'''
df_booking_lead_time = pd.read_sql_query(sql_query, conn)
median_booking_lead_time = df_booking_lead_time['booking_lead_time_days'].median()
print(f'Median booking lead time: {median_booking_lead_time:.2f} days')
```

```python
#calculation of average lead time day for each fare condition
sql_query = '''SELECT 
    tf.fare_conditions,
AVG(
        julianday(substr(f.scheduled_departure, 1, 19)) - julianday(substr(b.book_date, 1, 19))
    ) AS avg_lead_time_days,
    COUNT(*) AS total_tickets
FROM ticket_flights tf
JOIN tickets t 
    ON tf.ticket_no = t.ticket_no
JOIN bookings b
    ON t.book_ref = b.book_ref
JOIN flights f 
    ON tf.flight_id = f.flight_id
GROUP BY tf.fare_conditions
ORDER BY total_tickets DESC
'''
df_avg_booking_lead_time_fare_conditions = pd.read_sql_query(sql_query, conn)
display(df_avg_booking_lead_time_fare_conditions)
```

```python
#percentage of occupied seals for each fare condition
sql_query = '''SELECT 
    tf.fare_conditions,
    COUNT(*) AS occupied_seats,
    COUNT(*) * 100.0 / SUM(COUNT(*)) OVER() AS percentage
FROM boarding_passes bp
JOIN ticket_flights tf
    ON bp.ticket_no = tf.ticket_no
   AND bp.flight_id = tf.flight_id
GROUP BY tf.fare_conditions
ORDER BY percentage DESC'''

df_fare_conditions_occupied_seats = pd.read_sql_query(sql_query, conn)
display(df_fare_conditions_occupied_seats)
```

```python
#calculation of median lead time day for each fare condition
sql_query = '''SELECT 
    tf.fare_conditions,
(
        julianday(substr(f.scheduled_departure, 1, 19)) - julianday(substr(b.book_date, 1, 19))
    ) AS lead_time_days
FROM ticket_flights tf
JOIN tickets t 
    ON tf.ticket_no = t.ticket_no
JOIN bookings b
    ON t.book_ref = b.book_ref
JOIN flights f 
    ON tf.flight_id = f.flight_id
'''
df_booking_lead_time_fare_conditions = pd.read_sql_query(sql_query, conn)
df_median_booking_lead_time_fare_conditions = df_booking_lead_time_fare_conditions.groupby('fare_conditions')['lead_time_days'].median().reset_index()
df_median_booking_lead_time_fare_conditions['median lead time day'] = df_median_booking_lead_time_fare_conditions['lead_time_days']
df_median_booking_lead_time_fare_conditions = df_median_booking_lead_time_fare_conditions.drop(columns='lead_time_days')
display(df_median_booking_lead_time_fare_conditions)
```

```python
#average price tickets by booking time distribution
sql_query = '''
SELECT 
    CASE 
        WHEN julianday(substr(f.scheduled_departure,1,19)) - julianday(substr(b.book_date,1,19)) <= 3 
            THEN 'Last Minute (0-3 Tage)'
        WHEN julianday(substr(f.scheduled_departure,1,19)) - julianday(substr(b.book_date,1,19)) <= 14 
            THEN 'Kurzfristig (4-14 Tage)'
        WHEN julianday(substr(f.scheduled_departure,1,19)) - julianday(substr(b.book_date,1,19)) <= 30 
            THEN 'Mittelfristig (15-30 Tage)'
        ELSE 'Langfristig (>30 Tage)'
    END AS booking_window,
    
    COUNT(*) AS total_tickets,
    AVG(tf.amount) AS avg_ticket_price
FROM ticket_flights tf
JOIN tickets t 
    ON tf.ticket_no = t.ticket_no
JOIN bookings b
    ON t.book_ref = b.book_ref
JOIN flights f 
    ON tf.flight_id = f.flight_id
GROUP BY booking_window
ORDER BY avg_ticket_price DESC
'''
df_booking_time_distribution = pd.read_sql_query(sql_query, conn)
display(df_booking_time_distribution)
```

```python
#average price tickets by booking time distribution according to fare conditions
sql_query = '''SELECT 
    tf.fare_conditions,
    CASE 
        WHEN julianday(substr(f.scheduled_departure,1,19)) - julianday(substr(b.book_date,1,19)) <= 3 
            THEN 'Last Minute (0-3 Tage)'
        WHEN julianday(substr(f.scheduled_departure,1,19)) - julianday(substr(b.book_date,1,19)) <= 14 
            THEN 'Kurzfristig (4-14 Tage)'
        WHEN julianday(substr(f.scheduled_departure,1,19)) - julianday(substr(b.book_date,1,19)) <= 30 
            THEN 'Mittelfristig (15-30 Tage)'
        ELSE 'Langfristig (>30 Tage)'
    END AS booking_window,
    COUNT(*) AS total_tickets,
    AVG(tf.amount) AS avg_ticket_price
FROM ticket_flights tf
JOIN tickets t ON tf.ticket_no = t.ticket_no
JOIN bookings b ON t.book_ref = b.book_ref
JOIN flights f ON tf.flight_id = f.flight_id
GROUP BY tf.fare_conditions, booking_window
ORDER BY booking_window, total_tickets DESC'''

df_booking_time_distribution_fare_condition = pd.read_sql_query(sql_query, conn)
display(df_booking_time_distribution_fare_condition)
```
