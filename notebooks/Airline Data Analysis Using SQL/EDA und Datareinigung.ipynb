```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import sqlite3
pd.set_option('display.precision', 2) 
```

```python
#connection to databank
conn = sqlite3.connect("../../data/raw/airlines-dataset/travel.sqlite")
cursor = conn.cursor()
cursor.execute("SELECT name FROM sqlite_master WHERE type='table'")
tabels = cursor.fetchall()

print(tabels)
```

```python
for tabel in tabels:
    table_name = tabel[0]
    print('table name:', table_name)
    df_tabel = pd.read_sql_query(f"SELECT * FROM {table_name}", conn)
    display(df_tabel)
    print('\n')
```

```python
#closing the connection to databank
conn.close()
```

```python
#Die explorative Datenanalyse (EDA)
for tabel in tabels:
    print('name of table:', tabel)
    display(
        "Shape",
        df_tabel.shape,
    "Description",
    df_tabel.describe().round(2).T,
    "Duplicates",
    df_tabel.duplicated().sum(),
    )


    overview = pd.DataFrame(
        {
            "Data Types": df_tabel.dtypes,
            "Missing Values": df_tabel.isnull().sum(),
            "Unique Values": df_tabel.nunique(),
            "Sample Values": [df_tabel[col].sample(3).tolist() for col in df_tabel.columns]
        })
    display(overview)
    print('missing values')
    display(df_tabel.isna())
    print('\n'*2)
```
