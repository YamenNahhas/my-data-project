```python
def logistic_prediction(df: pd.DataFrame, feature: str, target: str) -> pd.Series:

    """
    return datafram series for the predicted target
    
    Parameters:
    -----------
    df : Datafram 
    feature : str (column name for independent Variable)
    target : str (column name for target variable(0/1))
    """

    model = smf.logit(formula=f'{target} ~ {feature}', data=df)
    results = model.fit(disp=False)

    value_range = np.linspace(df[feature].min(), df[feature].max(), 300)
    value_range_df = pd.DataFrame(value_range)
    value_range_df.columns=[feature]

    predicted_target = results.predict(value_range_df)
    return value_range, predicted_target
```

```python
ipytest.autoconfig() # Konfiguriert ipytest mit sinnvollen Standardeinstellungen
#test the function logistic_prediction
def test_logistic_prediction_basic():
    # Arrange: simple Test-DataFrame
    df = pd.DataFrame({
        "x": np.linspace(0, 10, 50),
    })
    # binäres Target mit klarer Beziehung
    df["y"] = (df["x"] > 5).astype(int)

    # Act
    value_range, predicted_target = logistic_prediction(
        df=df,
        feature="x",
        target="y"
    )

    # Assert: Search types
    assert isinstance(value_range, np.ndarray)
    assert isinstance(predicted_target, (pd.Series, np.ndarray))

    # Assert: Search the length
    assert len(value_range) == 300
    assert len(predicted_target) == 300

    # Assert: Value ranges of logistic regression
    assert np.all(predicted_target >= 0)
    assert np.all(predicted_target <= 1)


ipytest.run('-vv')
```
